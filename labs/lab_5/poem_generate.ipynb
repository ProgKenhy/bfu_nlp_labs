```python
import pandas as pd
import random

subjects = [
    "осень", "зима", "весна", "лето", "ночь", "день", "ветер", "дождь",
    "солнце", "луна", "небо", "сердце", "душа", "любовь"
]

verbs = [
    "шепчет", "молчит", "горит", "плывёт", "зовёт", "уходит",
    "возвращается", "скрывается", "сияет", "дрожит"
]

objects = [
    "в тишине", "во мгле", "вдали", "рядом", "внутри меня",
    "над землёй", "в облаках", "в сердце моём"
]

endings = [
    "тихо", "нежно", "снова", "печально", "медленно", "вечно"
]

def make_line():
    return f"{random.choice(subjects)} {random.choice(verbs)} {random.choice(objects)} {random.choice(endings)}"

def make_poem():
    lines = []
    for _ in range(random.randint(4, 8)):
        lines.append(make_line())
    return "\n".join(lines)

poems = [make_poem() for _ in range(5000)]

df = pd.DataFrame({'text': poems})
df.to_csv('poems.csv', index=False, encoding='utf-8')

print("✅ poems.csv создан (5000 стихов)")
```
